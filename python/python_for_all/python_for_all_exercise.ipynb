# Exercício 1: 

Escreva um programa simples para simular a operação do
comando grep em Unix. Peça ao usuário para entrar com uma expressão
regular e conte o número de linhas que se igualam à expressão digitada:

```bash
$ python grep.py
Digite uma expressão regular: ^Autor
mbox.txt teve 1798 linhas que se igualam a ^Autor
```

````python
import re

file = open("mbox.txt")
count = 0;

expression = input("insira a expressão regular: ")

if len(expression) > 0:
    for line in file:
        line = line.strip()
        value = re.findall(expression, line)
        
        if len(value) > 0:
           count += 1
          

print(f"{file.name} teve {count} linhas que se igualam a {expression}")
        
````

# Exercício 2: 
Escreva um programa que procure por uma linha na forma:
*Nova Revisão: 39772*

Extraia o número de cada linha usando uma expressão regular e o
método findall(). Compute o valor médio dos números e mostre-o.

```txt
Arquivo de entrada: mbox.txt
38444.0323119

Arquivo de entrada: mbox-short.txt
39756.9259259
```

````python
import re

name_file = input("Arquivo de entrada: ")

try:
    file = open(name_file)
except:
    print("Arquivo inexistente")
    import sys
    sys.exit()

total = 0
count = 0

for line in file:
    line = line.strip()
    value = re.findall("^New .*: ([0-9]+)", line)
    if len(value) > 0:
        count += 1
        total += int(value[0])
media = total / count
print(media)
````
