```python
numbers = []

while True:
    value = input("Insira o número: ")
    
    if value.lower() == "done":
        break

    try:
        number = float(value)
    except:
        print("Insira um valor númerico")

    numbers.append(number)

average = sum(numbers) / len(numbers)
print(f"Average: {average:.2f}")
```

```python
def modify_list(l):
    del(l[1], l[len(l) - 1])


def cut(l):
    return l[1:len(l) - 1]
```

```python
list_work = list(range(6))

modify_list(list_work)
list_work

print(cut(list_work))
```

```python
file = open("romeo.txt")

words = [];

for line in file:
    words_in_line = line.split()
    for word in words_in_line:
        if word in words: continue

        words.append(word)

words.sort()
            
print(words)
```

```python
numbers = []

while True:
    value = input("Insira o número: ")
    if value.lower() == "done": break

    try:
        value = float(value)
    except:
        print("Insira um valor válido")
        continue

    numbers.append(value)

print(f"Max: {max(numbers)}")
print(f"Min: {min(numbers)}")
```

```python
file = open("words.txt")

dictionary = dict()

for line in file:
    words = line.split()
    for word in words:
        if word in dictionary: continue

        dictionary[word] = "Magic"

print(dictionary)
    
```

```python
word = "programação"
d = dict()

for letter in word:
    d[letter] = d.get(letter, 0) + 1

print(d)
```

```python
# count words

file_name = input("Introduza o nome do arquivo: ")

file = open(file_name)
dictionary = dict()

for line in file:
    words = line.split()
    for word in words:
        dictionary[word] = dictionary.get(word, 0) + 1

for key, value in dictionary.items():
    print(f"{key} : {value}")
```

```python
import string

ile_name = input("Introduza o nome do arquivo: ")

file = open(file_name)
dictionary = dict()

for line in file:
    line = line.rstrip()
    line = line.translate(line.maketrans("", "", string.punctuation))
    line = line.lower()
    words = line.split()
    for word in words:
        dictionary[word] = dictionary.get(word, 0) + 1

for key, value in dictionary.items():
    print(f"{key} : {value}")
```
