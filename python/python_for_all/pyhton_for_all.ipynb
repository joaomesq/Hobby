# Expressões regulares

```python
import re

file = open("mbox.txt")

for line in file:
    line = line.strip()
    if re.search("^From:.+@", line):
        print(line)
```

```python
import re

s = 'A message from csev@umich.edu to cwen@iupui.edu about meeting @2PM'
ls = re.findall("\\S+@\\S+", s)
print(ls)
```
